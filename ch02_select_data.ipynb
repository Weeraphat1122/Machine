# บทที่ 2 · เข้าถึงและคัดเลือกข้อมูล

ข้อมูล 12 แถวดูง่าย แต่ข้อมูลจริงมักมีหลายพันแถว — สกิลสำคัญของนักวิเคราะห์คือ "หยิบเฉพาะส่วนที่ต้องการ" ซึ่งทำได้ 3 แนวทาง: เลือก**คอลัมน์** เลือก**แถว** และคัดกรอง**ตามเงื่อนไข**

เราจะใช้ข้อมูลสภาพอากาศจริงชุดเดิม (`data/thai_monthly_climate_1991_2020.csv` — รายละเอียดอ่านจากการ์ดข้อมูลบทที่ 1)

```python
import pandas as pd

df = pd.read_csv("data/thai_monthly_climate_1991_2020.csv")

# เลือกคอลัมน์: ชื่อเดียว -> Series / หลายชื่อ (วงเล็บซ้อน 2 ชั้น) -> DataFrame
print(df["month_th"])                    # Series
print()
print(df[["month_th", "cm_mean_c"]])      # DataFrame
```

## เลือกแถว: `loc` กับ `iloc`

สองคำสั่งนี้มักสับสน จำง่าย ๆ: `loc` = **lo**cate ตาม "ชื่อป้ายกำกับ" · `iloc` = **i**nteger ตาม "ลำดับตัวเลข" เริ่มนับจาก 0 และการตัดช่วงต่างกัน:

| คำสั่ง | อ้างอิงด้วย | ตัวอย่าง | ได้แถว |
|---|---|---|---|
| `loc` | ชื่อ/ป้ายกำกับ | `df.loc[1:3]` | 1, 2, 3 *(รวมเส้นขอบ)* |
| `iloc` | ลำดับตำแหน่ง (เริ่ม 0) | `df.iloc[1:3]` | 1, 2 *(ไม่รวมเส้นขอบ เหมือน list)* |

```python
# เลือกแถว + คอลัมน์พร้อมกัน
print(df.loc[3])                                # เดือนเมษายน (index ชื่อ 3)
print()
print(df.iloc[0])                                # แถวแรกสุด (ลำดับ 0)
print()
print(df.loc[1:3, ["month_th", "cm_mean_c"]])    # แถว 1-3 *รวม 3* + บางคอลัมน์
print()
print(df.iloc[4:6, 0:3])                         # ตำแหน่ง 4-5 (ไม่รวม 6) + คอลัมน์ 0-2
```

## คัดกรองด้วยเงื่อนไข — หัวใจของการวิเคราะห์

รูปแบบคือ `df[เงื่อนไข]` โดยเงื่อนไขคือการเปรียบเทียบทั้งคอลัมน์ เช่น `df["cm_rain_mm"] > 150` จะได้ True/False ทีละแถว แล้ว pandas คืนเฉพาะแถวที่เป็น True

> ⚠️ หลายเงื่อนไขต้องใช้ `&` (และ) / `|` (หรือ) และ **ครอบวงเล็บแต่ละเงื่อนไขเสมอ** — ใช้ `and`/`or` จะ error ทันที

```python
# คัดกรองจากข้อมูลจริง
rainy = df[df["cm_rain_mm"] > 150]          # เดือนที่ฝนเชียงใหม่เกิน 150 มม.
print("เดือนที่เชียงใหม่ฝน > 150 มม.:")
print(rainy[["month_en", "cm_rain_mm"]])
print()

hot_bkk = df[df["bkk_mean_c"] > 29.5]       # เดือนที่กทม. ร้อนเกิน 29.5°C
print("เดือนที่กทม. ร้อน > 29.5°C:", list(hot_bkk["month_en"]))
print()

# หลายเงื่อนไข: เดือนที่เชียงใหม่ "เย็นกว่า 25" และ "ฝนน้อยกว่า 50"
cool_dry = df[(df["cm_mean_c"] < 25) & (df["cm_rain_mm"] < 50)]
print("เดือนเย็นและแห้งของเชียงใหม่:", list(cool_dry["month_en"]))
```

## เพิ่มและลบคอลัมน์

การเพิ่มคอลัมน์ทำได้ง่ายเหมือนเพิ่ม key ใน dict — ตั้งชื่อใหม่แล้วชี้ค่า โดยค่านั้นคำนวณจากคอลัมน์เดิมทั้งคอลัมน์ได้เลย (vector คำนวณพร้อมกันทุกแถว)

```python
# เพิ่มคอลัมน์: อุณหภูมิเฉลี่ยของ 3 เมือง / ฝนรวมของ 3 เมือง
df["avg_mean_c"] = (df["cm_mean_c"] + df["bkk_mean_c"] + df["phkt_mean_c"]) / 3
df["rain_total_mm"] = df["cm_rain_mm"] + df["bkk_rain_mm"] + df["phkt_rain_mm"]

print(df[["month_en", "avg_mean_c", "rain_total_mm"]].round(2))
print()

# เรียงตามฝนรวมมาก -> น้อย แล้วดู 3 อันดับ
top_rain = df.sort_values("rain_total_mm", ascending=False)
print("3 เดือนที่ฝนมากที่สุด (รวม 3 เมือง):")
print(top_rain[["month_en", "rain_total_mm"]].head(3))

# ลบคอลัมน์ (drop คืนตารางใหม่ ไม่แก้ตารางเดิม)
# df = df.drop(columns=["rain_total_mm"])   # ลองลบคอมเมนต์แล้วรันดูได้
```

## 🎯 ภารกิจบทที่ 2 (ส่งคำตอบใน Moodle)

จากข้อมูลค่าปกติ 30 ปี **เดือนที่เชียงใหม่ "ร้อนที่สุด" คือเดือนที่เท่าไร** (ตอบเป็นเลขเดือน 1–12)

*ใบ้: ใช้ `.idxmax()` หาตำแหน่งที่ค่ามากที่สุดของ Series แล้วดึงเลขเดือนจากคอลัมน์ `month`*

```python
# ภารกิจ 2: เดือนที่เชียงใหม่ร้อนที่สุด (ตอบเลขเดือน 1-12)
hottest = df.loc[df["cm_mean_c"].idxmax()]
print("เดือนที่ร้อนที่สุดของเชียงใหม่:", hottest["month_en"])
print("คำตอบภารกิจบทที่ 2 =", int(hottest["month"]))
```

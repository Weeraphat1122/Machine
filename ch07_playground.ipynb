# บทที่ 7 · สนามซ้อมโค้ด — โจทย์ท้าทาย

พื้นที่อิสระ ไม่มีเฉลยใน notebook — ทุกอย่างที่เรียนมาใช้ได้ที่นี่หมด เขียนโค้ดของตัวเองลงในเซลล์ว่างแล้วรันเลย

**โจทย์ท้าทาย 4 ข้อ** (ใช้ข้อมูลกาแฟ NYC จากการ์ดข้อมูลบทที่ 5):

1. **โจทย์ 1** — เรียง 10 อันดับ `product_detail` ที่ *จำนวนธุรกรรม* มากที่สุด (บทที่ 2–3)
2. **โจทย์ 2** — ช่วงเวลาไหนของวันที่ลูกค้าเยอะที่สุด? แยกเป็น "ชั่วโมง" จาก `transaction_time` (ใบ้: `pd.to_datetime(..., format="%H:%M:%S").dt.hour`) (บทที่ 2–3)
3. **โจทย์ 3** — วาดกราฟแท่งจำนวนธุรกรรมแยกตามชั่วโมง พร้อมชื่อกราฟและชื่อแกน (บทที่ 4–5)
4. **โจทย์ 4 (ท้าทาย)** — เพิ่มคอลัมน์ `price_level`: Cheap (< $4), Medium ($4–6), Premium (> $6) จาก `unit_price` แล้วนับว่าแต่ละกลุ่มมีธุรกรรมกี่รายการ (ใบ้: ใช้ `pd.cut()` หรือ list comprehension + for/if)

```python
# พื้นที่เริ่มต้น: โหลดข้อมูล + ตัวแปรช่วย (แก้/เพิ่มอะไรก็ได้)
import pandas as pd
import matplotlib.pyplot as plt

coffee = pd.read_csv("data/coffee_sales_nyc_2023.csv.gz")
coffee["revenue"] = coffee["transaction_qty"] * coffee["unit_price"]
coffee["month"] = pd.to_datetime(coffee["transaction_date"], format="%d-%m-%Y").dt.month
coffee["hour"] = pd.to_datetime(coffee["transaction_time"], format="%H:%M:%S").dt.hour

print(coffee.head(2))
print("ขนาดข้อมูล:", coffee.shape)

# เขียนโค้ดโจทย์ 1-4 ต่อจากบรรทัดนี้...

# ตัวอย่างเฉลยโจทย์ 2: ชั่วโมงที่ธุรกรรมเยอะที่สุด
by_hour = coffee["hour"].value_counts().sort_index()
print("ชั่วโมงที่ธุรกรรมเยอะที่สุด:", by_hour.idxmax(), "น. =", by_hour.max(), "ธุรกรรม")
```

## 🎯 ภารกิจบทที่ 7 (ส่งคำตอบใน Moodle)

ร้านเปิดมา 181 วันเต็ม (1 ม.ค. – 30 มิ.ย. 2023) **เฉลี่ยแล้วมีธุรกรรมกี่รายการต่อวัน** (ปัดเป็นจำนวนเต็ม)

*ใบ้: จำนวนธุรกรรมทั้งหมด ÷ จำนวนวันที่ไม่ซ้ำ (`nunique()` ของคอลัมน์วันที่)*

```python
# ภารกิจ 7: ธุรกรรมเฉลี่ยต่อวัน (ปัดเป็นจำนวนเต็ม)
n_days = coffee["transaction_date"].nunique()
per_day = round(len(coffee) / n_days)
print("จำนวนวันที่เปิดร้าน:", n_days, "วัน")
print("คำตอบภารกิจบทที่ 7 =", per_day)
```

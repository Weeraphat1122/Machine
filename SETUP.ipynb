## วิธีใช้ใน VS Code
1. เปิด VS Code → `File > Open Folder` → เลือกโฟลเดอร์ **student_files** ที่แตกไฟล์แล้ว
2. เปิดไฟล์ `SETUP.ipynb` นี้
3. กดเซลล์แรก แล้วกดปุ่ม **▶ Run Cell** (รันเซลล์) ที่มุมซ้ายของเซลล์
4. ครั้งแรก VS Code จะถามให้ **เลือก Python Kernel** — เลือกสภาพแวดล้อมที่เราสร้างไว้ตอนติดตั้ง (ดูคู่มือใน Moodle บทที่ 0)
5. รันเซลล์จากบนลงล่างตามลำดับเสมอ — ตัวแปรจะจำไปใช้เซลล์ถัดไป

> ❓ ถ้าขึ้น error ว่า `No module named 'pandas'` ให้กลับไปดูขั้นตอนติดตั้งใน Moodle บทที่ 0 (เรื่อง venv + `pip install -r requirements.txt`)

```python
# ตรวจว่าติดตั้งครบ 3 ไลบรารีหลัก
import sys
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt

print("Python      :", sys.version.split()[0])
print("pandas      :", pd.__version__)
print("numpy       :", np.__version__)
print("matplotlib  :", matplotlib.__version__)
print()
print("ทุกไลบรารีพร้อมใช้งาน ✅" if pd.__version__ and matplotlib.__version__ else "มีบางอย่างผิดปกติ")
```
